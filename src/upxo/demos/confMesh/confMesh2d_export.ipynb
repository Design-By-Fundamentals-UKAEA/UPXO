# Conformal 2D mesh: visualisation and Abaqus export

Uses existing UPXO modules only:

- `upxo.meshing.gsmesh2d.mesh_gs` / `visualize_gs_mesh`
- `upxo.viz.meshviz.plot_conformal_2d_by_grain` (via `confMesh2dGMSH.plot_by_grain`)
- `upxo.meshing.writer_ABQ.export_confmesh2d_inp` (via `confMesh2dGMSH.export_abaqus_inp`)
- `upxo.meshing.writer_ABQ.summarize_inp`

```python
from pathlib import Path
from shapely.geometry import box

from upxo.meshing.gsmesh2d import mesh_gs, visualize_gs_mesh
from upxo.meshing.writer_ABQ import summarize_inp
```

```python
cells = {1: box(0, 0, 2, 1), 2: box(2, 0, 4, 1), 3: box(0, 1, 4, 2)}
result = mesh_gs(cells, mesh_size_gb=0.25, mesh_size_bulk=0.5,
                 mesh_algo=6, recombine_to_quads=False)
m = result['mesher']
m.form_elsets_gmsh()
m.build_boundary_nsets()
m.build_gb_nset()
m.validation_report
```

## Visualisation

```python
fig, ax = visualize_gs_mesh(result, figsize=(8, 3.5), dpi=130, show_nsets=True)
fig
```

## Abaqus INP (`plane='stress'` → CPS3, `'strain'` → CPE3)

```python
out = Path.cwd() / 'confMesh2d_export_out'
inp_cps = m.export_abaqus_inp(out / 'rve_cps3.inp', plane='stress')
inp_cpe = m.export_abaqus_inp(out / 'rve_cpe3.inp', plane='strain')
summarize_inp(inp_cps), inp_cps, inp_cpe
```
